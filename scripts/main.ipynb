```python
def get_ai_response(
    user_prompt: str, 
    system_prompt: str = "You are a helpful assistant.", 
    model_name: str = "kimi-k2.5", 
    api_key: str = "",
    base_url: str = "https://helixmind.online/v1"
):
    """
    Calls an OpenAI-compatible API endpoint.
    """
    
    # Initialize the client with configurable base_url
    client = OpenAI(
        api_key=api_key,
        base_url=base_url
    )

    try:
        response = client.chat.completions.create(
            model=model_name,
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": user_prompt}
            ],
            temperature=0.7  # Optional: controls randomness
        )
        
        return response.choices[0].message.content

    except Exception as e:
        return f"An error occurred: {e}"
```

```python
def generate_and_show_image(
    user_prompt: str, 
    model_name: str = "image-1", 
    api_key: str = "",
    base_url: str = "https://helixmind.online/v1/images/generations"
):
    headers = {
    "Authorization": f"Bearer {api_key}",
    "Content-Type": "application/json"
    }

    payload = {
        "model": model_name,  # Options: FLUX.1-pro, FLUX.1-schnell, etc.
        "prompt": user_prompt,
        "n": 1,              # Number of images to generate
        "size": "1024x1024",  # Check specific model docs for supported sizes
        "quality":"standard"
    }
    
    try:
        response = requests.post(base_url, headers=headers, json=payload)
        response.raise_for_status()
        
        # The response usually contains a URL or base64 data
        data = response.json()
        return data

    except Exception as e:
        return f"An error occurred: {e}"
```

```python
# --- Configuration & Execution ---
if __name__ == "__main__":
    # 1. Configuration
    MY_MODEL = "image-1"
    USER_QUERY = """
    A detailed cinematic photograph capturing a quiet, sun-drenched corner of an ancient stone courtyard. A single, sharp beam of natural light pierces through a high, arched window, illuminating dancing dust motes in the air. In the soft shadows, delicate and intricate spider webs cling to the weathered textures of the carved pillars. A rustic wooden ledge and cracked earth. The atmosphere is one of heavy, dignified silence and profound stillness. The color palette is composed of warm ochres, deep umbers, and soft earth tones. The focus is sharp on the interplay of light and shadow, evoking a sense of waiting and the quiet, heavy weight of unsaid words. 4k resolution, natural lighting, soulful and serene.


    """
    
    # 2. Call the function
    result = generate_and_show_image(
        user_prompt=USER_QUERY,
        model_name=MY_MODEL
    )
    
    print(result)
    
    if result:
        img_response = requests.get(result['data'][0]['url'])
        img = Image.open(BytesIO(img_response.content))

        display(img)
    else:
        print(result)

    # img.save("dalle_image.png")
```

```python
from docx import Document

doc = Document('मौन के हस्ताक्षर.docx')
pages = []
current_page_text = []

for para in doc.paragraphs:
    # Check if this paragraph contains a manual page break
    # 'w:br' with type 'page' represents a hard break
    if 'w:br' in para._element.xml and 'type="page"' in para._element.xml:
        pages.append("\n".join(current_page_text))
        current_page_text = [para.text] # Start new page
    else:
        current_page_text.append(para.text)

# Don't forget the last page
if current_page_text:
    pages.append("\n".join(current_page_text))

# Now 'pages' is an array where pages[0] is Page 1, etc.
print(f"Total pages captured: {len(pages)}")
print(pages)
```
